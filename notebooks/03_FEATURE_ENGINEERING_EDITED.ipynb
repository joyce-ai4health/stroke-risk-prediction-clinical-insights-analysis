# Section 9: Encode Binary Categorical Variables

## Business Question

How can we convert binary categorical variables into numerical values that machine learning algorithms can understand?

---

## Why This Step Matters

Machine learning algorithms cannot directly interpret text values such as **Male**, **Female**, **Yes**, **No**, **Urban**, or **Rural**.

Label Encoding converts these text categories into numbers while preserving all the information.

Since these variables have only two categories, Label Encoding is the most appropriate technique.

The variables to be encoded are:

- gender
- ever_married
- Residence_type

```python
# =====================================================
# SECTION 10: LABEL ENCODING
# =====================================================

# Create a LabelEncoder object.
#
# This object converts text values into numerical values.
# Example:
#
# Male   -> 1
# Female -> 0
#
# The exact numbers assigned are determined automatically.

label_encoder = LabelEncoder()

# -----------------------------------------------
# Encode the Gender column
# -----------------------------------------------

# Convert Male/Female into numerical values.

df["gender"] = label_encoder.fit_transform(df["gender"])

# -----------------------------------------------
# Encode Ever Married
# -----------------------------------------------

# Convert:
#
# Yes -> 1
# No  -> 0

df["ever_married"] = label_encoder.fit_transform(df["ever_married"])

# -----------------------------------------------
# Encode Residence Type
# -----------------------------------------------

# Convert:
#
# Urban
# Rural
#
# into numerical values.

df["Residence_type"] = label_encoder.fit_transform(df["Residence_type"])

# Display the first five rows to verify the changes.

df.head()
```

```python
# =====================================================
# SECTION 11: VERIFY LABEL ENCODING
# =====================================================

# Display only the encoded columns.
#
# This helps confirm that the binary categorical
# variables have been converted successfully.

df[
    [
        "gender",
        "ever_married",
        "Residence_type"
    ]
].head()
```

## Interpretation

The binary categorical variables have now been successfully converted into numerical values.

These transformations allow machine learning algorithms to process the variables while preserving the original information contained in each category.

Since each variable contains only two categories, Label Encoding is an appropriate and efficient encoding technique.

# Section 12: One-Hot Encode Multi-category Variables

## Business Question

How can we convert categorical variables with multiple categories into a format suitable for machine learning without introducing artificial ordering?

---

## Why This Step Matters

Unlike binary variables, columns such as **work_type** and **smoking_status** contain several categories.

If Label Encoding were applied to these variables, the model might incorrectly assume that one category is greater than another.

For example:

Private = 0

Government Job = 1

Self-employed = 2

This would incorrectly introduce an order that does not exist.

Instead, One-Hot Encoding creates a separate binary column for each category, allowing the model to treat every category independently.

```python
# =====================================================
# SECTION 13: ONE-HOT ENCODING
# =====================================================

# Unlike binary variables, work_type and smoking_status
# contain multiple categories.
#
# Applying Label Encoding to these variables would
# incorrectly introduce an order between categories.
#
# One-Hot Encoding creates a separate binary column
# for each category while preserving all information.

# ----------------------------------------------
# One-Hot Encode work_type
# ----------------------------------------------

# Create dummy variables for each work type.
# drop_first=False keeps every category so that
# no information is lost.

work_type_encoded = pd.get_dummies(
    df["work_type"],
    prefix="work_type",
    drop_first=False
)

# ----------------------------------------------
# One-Hot Encode smoking_status
# ----------------------------------------------

# Create dummy variables for each smoking status.

smoking_encoded = pd.get_dummies(
    df["smoking_status"],
    prefix="smoking_status",
    drop_first=False
)

# ----------------------------------------------
# Add the encoded columns back to the dataset
# ----------------------------------------------

df = pd.concat(
    [df, work_type_encoded, smoking_encoded],
    axis=1
)

# ----------------------------------------------
# Remove the original categorical columns
# ----------------------------------------------

# These columns are no longer needed because
# their information is now stored in the newly
# created binary columns.

df.drop(
    columns=[
        "work_type",
        "smoking_status"
    ],
    inplace=True
)

# Display the first five rows to verify the changes.

df.head()
```

```python
# =====================================================
# SECTION 14: VERIFY ONE-HOT ENCODING
# =====================================================

# Display only the newly created columns
# to confirm that the encoding was successful.

encoded_columns = [

    column

    for column in df.columns

    if column.startswith("work_type_")

    or column.startswith("smoking_status_")

]

df[encoded_columns].head()
```

## Interpretation

The multi-category variables have now been successfully transformed using One-Hot Encoding.

Instead of assigning numerical labels that could imply an artificial order, each category is represented by its own binary column.

This allows machine learning models to learn the influence of each category independently while avoiding incorrect assumptions about relationships between categories.

At the same time, the original categorical columns have been removed because their information is now fully represented by the newly created binary variables.

# Section 15: Create Age Band Feature

## Business Question

Can grouping patients into clinically meaningful age bands help the model better capture differences in stroke risk?

---

## Why This Step Matters

Although the original age variable will remain in the dataset, creating age bands may help machine learning algorithms recognise non-linear patterns that are difficult to capture using age alone.

Based on the exploratory data analysis, stroke risk increased substantially among older patients.

The following age bands will therefore be created:

- Less than 40 years
- 40 to 60 years
- Above 60 years

Importantly, the original continuous **age** variable will remain in the dataset, following the mentor's recommendation to preserve important clinical measurements.

```python
# =====================================================
# SECTION 16: CREATE AGE BAND FEATURE
# =====================================================

# Create age categories based on clinically
# meaningful age groups identified during EDA.

df["age_band"] = pd.cut(

    df["age"],

    bins=[0, 40, 60, 120],

    labels=[

        "<40",

        "40-60",

        "60+"

    ]

)

# Display a sample of the new feature.

df[
    [
        "age",
        "age_band"
    ]
].head(10)
```

## Interpretation

A new categorical feature named **age_band** has been created.

This engineered feature groups patients into clinically meaningful age categories while preserving the original continuous age variable.

Keeping both variables allows predictive models to benefit from precise numerical age values as well as broader age-related risk patterns identified during exploratory data analysis.

# Section 17: Create Glucose Risk Tier

## Business Question

Can grouping patients according to their blood glucose levels help the predictive model identify clinically meaningful patterns associated with stroke risk?

---

## Why This Step Matters

Average glucose level is already available as a continuous numerical variable.

However, clinicians often interpret glucose measurements using recognised risk categories rather than exact numbers alone.

Creating glucose risk tiers allows the model to learn both:

- The exact glucose measurement (continuous variable).
- The patient's clinical risk category (engineered feature).

Following the mentor's recommendation, the original **avg_glucose_level** variable will remain in the dataset.

The glucose categories are:

- **Normal:** Less than 100 mg/dL
- **Pre-diabetic:** 100–125 mg/dL
- **Diabetic:** Greater than 125 mg/dL

```python
# =====================================================
# SECTION 18: CREATE GLUCOSE RISK TIER
# =====================================================

# Create clinically meaningful glucose categories.
#
# We use pd.cut() to divide continuous glucose values
# into three clinical risk groups.

df["glucose_risk_tier"] = pd.cut(

    df["avg_glucose_level"],

    bins=[0, 100, 125, float("inf")],

    labels=[
        "Normal",
        "Pre-diabetic",
        "Diabetic"
    ]

)

# Display a sample of the original glucose values
# alongside the newly created glucose risk tier.

df[
    [
        "avg_glucose_level",
        "glucose_risk_tier"
    ]
].head(10)
```

## Interpretation

A new feature called **glucose_risk_tier** has been created.

This feature groups patients into clinically recognised glucose categories while preserving the original continuous glucose measurements.

Keeping both variables allows machine learning models to learn from both the exact glucose values and the broader clinical risk categories.

# Section 19: Create BMI Category

## Business Question

Can grouping Body Mass Index (BMI) into standard clinical categories improve the model's ability to identify stroke risk?

---

## Why This Step Matters

BMI is already available as a continuous variable.

However, healthcare professionals usually interpret BMI using standard clinical categories.

Creating BMI categories allows the model to capture obesity-related risk patterns while still preserving the original BMI measurements.

Following the mentor's recommendation, the original **bmi** variable will remain in the dataset.

The BMI categories are:

- Underweight (<18.5)
- Normal (18.5–24.9)
- Overweight (25–29.9)
- Obese (30+)

```python
# =====================================================
# SECTION 20: CREATE BMI CATEGORY
# =====================================================

# Create clinically meaningful BMI categories.
#
# These categories are based on standard WHO BMI
# classifications.

df["bmi_category"] = pd.cut(

    df["bmi"],

    bins=[0, 18.5, 25, 30, float("inf")],

    labels=[
        "Underweight",
        "Normal",
        "Overweight",
        "Obese"
    ]

)

# Display a sample of the original BMI values
# together with the newly created BMI category.

df[
    [
        "bmi",
        "bmi_category"
    ]
].head(10)
```

## Interpretation

A new feature called **bmi_category** has been created.

This feature groups patients according to standard clinical BMI classifications while preserving the original BMI measurements.

Retaining both variables provides the predictive model with detailed numerical information as well as clinically meaningful categories.
